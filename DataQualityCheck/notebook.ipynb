```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("Cleaned_DS_Jobs.csv")
data.head()
```

```python
data.shape
```

```python
missing = data.isnull().sum()
print(missing)
```

```python
null_count = data.isnull().sum()
missing_percentage = (null_count/ len(data))*100
print(missing_percentage)
```

```python
data.duplicated().sum()
```

```python
data[data.duplicated()]
```

```python
duplicate_count = data.duplicated().sum()
duplicate_percentage = (duplicate_count / len(data))*100
print(duplicate_percentage)
```

```python
data = data.drop_duplicates()
data.duplicated().sum()
```

```python
data_types = data.dtypes
print(data_types)
```

```python
data["max_salary"].describe()
```

```python
Q1 = data['max_salary'].quantile(0.25)
Q3 = data['max_salary'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[
    (data['max_salary'] < lower) |
    (data['max_salary'] > upper)
]

print(outliers)
```

```python
print(len(outliers))
print((len(outliers) / len(data)) * 100)
```

```python
unique_values = data.nunique()
```

```python
# Create report
quality_report = pd.DataFrame({
    'Data Type': data_types,
    'Missing Values': missing,
    'Missing %': missing_percentage.round(2),
    'Unique Values': unique_values
})

print("Total Rows:", len(data))
print("Total Columns:", len(data.columns))
print("Total Cells:", data.size)
print("Duplicate Rows:", duplicate_count)
print("Duplicate %:", round(duplicate_percentage, 2))

display(quality_report)
```
